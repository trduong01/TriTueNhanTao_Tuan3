```
#Bai1
import math

# Người chơi
HUMAN = 'O'
AI = 'X'
EMPTY = ' '

# Khởi tạo bàn cờ
def create_board():
    return [[EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY]]

# In bàn cờ
def print_board(board):
    for row in board:
        print("| " + " | ".join(row) + " |")
    print()

# Kiểm tra thắng thua
def check_winner(board):
    # Check rows
    for row in board:
        if row[0] == row[1] == row[2] != EMPTY:
            return row[0]

    # Check cols
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] != EMPTY:
            return board[0][col]

    # Check diagonals
    if board[0][0] == board[1][1] == board[2][2] != EMPTY:
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != EMPTY:
        return board[0][2]

    return None

# Kiểm tra full bàn cờ
def is_full(board):
    for row in board:
        if EMPTY in row:
            return False
    return True

# Minimax
def minimax(board, depth, is_maximizing):
    winner = check_winner(board)

    if winner == AI:
        return 1
    if winner == HUMAN:
        return -1
    if is_full(board):
        return 0

    # AI maximize
    if is_maximizing:
        best_score = -math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == EMPTY:
                    board[i][j] = AI
                    score = minimax(board, depth + 1, False)
                    board[i][j] = EMPTY
                    best_score = max(best_score, score)
        return best_score

    # Human minimize
    else:
        best_score = math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == EMPTY:
                    board[i][j] = HUMAN
                    score = minimax(board, depth + 1, True)
                    board[i][j] = EMPTY
                    best_score = min(best_score, score)
        return best_score

# Chọn nước đi tốt nhất cho AI
def best_move(board):
    best_score = -math.inf
    move = None

    for i in range(3):
        for j in range(3):
            if board[i][j] == EMPTY:
                board[i][j] = AI
                score = minimax(board, 0, False)
                board[i][j] = EMPTY

                if score > best_score:
                    best_score = score
                    move = (i, j)
    return move

# Game loop
def play_game():
    board = create_board()
    print("TicTacToe 3x3 – Minimax\n")
    print_board(board)

    while True:
        # Human move
        row = int(input("Nhập dòng (0-2): "))
        col = int(input("Nhập cột (0-2): "))

        if board[row][col] != EMPTY:
            print("Ô đã có người đánh!")
            continue

        board[row][col] = HUMAN

        print_board(board)
        if check_winner(board) == HUMAN:
            print("Bạn thắng!")
            break
        if is_full(board):
            print("Hòa!")
            break

        # AI move
        print("AI đang tính toán...")
        x, y = best_move(board)
        board[x][y] = AI

        print_board(board)
        if check_winner(board) == AI:
            print("AI thắng!")
            break
        if is_full(board):
            print("Hòa!")
            break


# Chạy game
play_game()
```

```
#Bai2
import math

HUMAN = 'O'
AI = 'X'
EMPTY = ' '

# Tạo bàn cờ
def create_board():
    return [[EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY],
            [EMPTY, EMPTY, EMPTY]]

# In bàn cờ
def print_board(board):
    for row in board:
        print("| " + " | ".join(row) + " |")
    print()

# Kiểm tra thắng
def check_winner(board):
    # hàng
    for row in board:
        if row[0] == row[1] == row[2] != EMPTY:
            return row[0]

    # cột
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] != EMPTY:
            return board[0][col]

    # chéo
    if board[0][0] == board[1][1] == board[2][2] != EMPTY:
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != EMPTY:
        return board[0][2]

    return None

# Kiểm tra bàn cờ đầy
def is_full(board):
    for row in board:
        if EMPTY in row:
            return False
    return True


# =============================
# ALPHA - BETA PRUNING
# =============================
def alphabeta(board, depth, alpha, beta, maximizing):
    winner = check_winner(board)

    # Điểm khi kết thúc
    if winner == AI:
        return 1
    if winner == HUMAN:
        return -1
    if is_full(board):
        return 0

    # Nước đi của AI (maximize)
    if maximizing:
        max_eval = -math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == EMPTY:
                    board[i][j] = AI
                    eval = alphabeta(board, depth + 1, alpha, beta, False)
                    board[i][j] = EMPTY
                    max_eval = max(max_eval, eval)
                    alpha = max(alpha, eval)

                    # Cắt tỉa
                    if beta <= alpha:
                        return max_eval

        return max_eval

    # Nước đi của người (minimize)
    else:
        min_eval = math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == EMPTY:
                    board[i][j] = HUMAN
                    eval = alphabeta(board, depth + 1, alpha, beta, True)
                    board[i][j] = EMPTY
                    min_eval = max(min_eval, eval)
                    beta = min(beta, eval)

                    # Cắt tỉa
                    if beta <= alpha:
                        return min_eval

        return min_eval


# =============================
# CHỌN NƯỚC ĐI TỐT NHẤT CHO AI
# =============================
def best_move(board):
    best_score = -math.inf
    move = None

    for i in range(3):
        for j in range(3):
            if board[i][j] == EMPTY:
                board[i][j] = AI
                score = alphabeta(board, 0, -math.inf, math.inf, False)
                board[i][j] = EMPTY

                if score > best_score:
                    best_score = score
                    move = (i, j)

    return move


# =============================
# CHƠI GAME
# =============================
def play_game():
    board = create_board()
    print("TicTacToe 3x3 – Alpha-Beta Pruning\n")
    print_board(board)

    while True:
        # Người chơi
        row = int(input("Nhập dòng (0-2): "))
        col = int(input("Nhập cột (0-2): "))

        if board[row][col] != EMPTY:
            print("Ô đã có người đánh!")
            continue

        board[row][col] = HUMAN
        print_board(board)

        if check_winner(board) == HUMAN:
            print("Bạn thắng!")
            break
        if is_full(board):
            print("Hòa!")
            break

        # AI
        print("AI đang tính toán...")
        x, y = best_move(board)
        board[x][y] = AI
        print_board(board)

        if check_winner(board) == AI:
            print("AI thắng!")
            break
        if is_full(board):
            print("Hòa!")
            break


# Chạy game
play_game()
```

```
#Bai3
import math

PLAYER = "X"
OPPONENT = "O"

def check_winner(board, N, K):
    """Kiểm tra thắng theo hàng, cột, chéo (K liên tiếp)."""
    lines = []


    for r in range(N):
        lines.append(board[r])


    for c in range(N):
        col = [board[r][c] for r in range(N)]
        lines.append(col)


    for start in range(N - K + 1):
        for c in range(N - K + 1):
            diag = [board[start + i][c + i] for i in range(K)]
            lines.append(diag)


    for start in range(N - K + 1):
        for c in range(K - 1, N):
            diag = [board[start + i][c - i] for i in range(K)]
            lines.append(diag)


    for line in lines:
        if all(cell == PLAYER for cell in line):
            return PLAYER
        if all(cell == OPPONENT for cell in line):
            return OPPONENT

    return None



def evaluate(board, N, K):
    win = check_winner(board, N, K)
    if win == PLAYER:
        return 100000
    if win == OPPONENT:
        return -100000
    return 0


def minimax(board, depth, is_max, N, K, max_depth):
    score = evaluate(board, N, K)

    if abs(score) == 100000:
        return score - depth if score > 0 else score + depth

    if depth == max_depth:
        return score  # giới hạn độ sâu

    if is_max:
        best = -math.inf
        for r in range(N):
            for c in range(N):
                if board[r][c] == " ":
                    board[r][c] = PLAYER
                    best = max(best, minimax(board, depth + 1, False, N, K, max_depth))
                    board[r][c] = " "
        return best

    else:  # Min
        best = math.inf
        for r in range(N):
            for c in range(N):
                if board[r][c] == " ":
                    board[r][c] = OPPONENT
                    best = min(best, minimax(board, depth + 1, True, N, K, max_depth))
                    board[r][c] = " "
        return best


def best_move_minimax(board, N, K, max_depth=2):
    best_score = -math.inf
    move = (-1, -1)

    for r in range(N):
        for c in range(N):
            if board[r][c] == " ":
                board[r][c] = PLAYER
                score = minimax(board, 0, False, N, K, max_depth)
                board[r][c] = " "
                if score > best_score:
                    best_score = score
                    move = (r, c)
    return move
```
